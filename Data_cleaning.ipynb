```python
import pandas as pd
import numpy as np


df = pd.read_csv('/content/drive/MyDrive/seasonal_agriculture_performance_dataset.csv')

# 1. Standardize column names
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')

# 2. Remove duplicate rows
df.drop_duplicates(inplace=True)

# 3. Handle missing values without chained inplace warnings
numerical_cols = df.select_dtypes(include=[np.number]).columns
categorical_cols = df.select_dtypes(include=['object']).columns

# Impute missing numerical values with median
for col in numerical_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())

# Impute missing categorical values with mode
for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

# 4. Feature Engineering for Tableau Analytics
# NPK Ratio / Soil Health Indicator
if 'nitrogen_kg_ha' in df.columns and 'phosphorus_kg_ha' in df.columns:
    df['np_ratio'] = df['nitrogen_kg_ha'] / (df['phosphorus_kg_ha'] + 0.001)

# Categorize Soil pH (Acidic, Neutral, Alkaline)
if 'soil_ph' in df.columns:
    df['soil_type_category'] = pd.cut(
        df['soil_ph'],
        bins=[0, 6.0, 7.5, 14.0],
        labels=['Acidic', 'Neutral', 'Alkaline']
    )
df.to_csv('/content/cleaned_seasonal_agriculture_performance.csv', index=False)
print("Cleaned file saved successfully as 'cleaned_seasonal_agriculture_performance.csv'")
```
